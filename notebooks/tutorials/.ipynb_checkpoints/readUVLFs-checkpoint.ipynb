```python
def param_wrapper(paramvector, accretion_model = accretion_model_DEF):
    'Input params are log10eps, log10Mc, a*, b*, for ease of sampling'
    log10eps, log10Mc, alphastar, betastar, sigmaUV, \
    dlog10eps, dlog10Mc, dalphastar, dbetastar, dsigmaUV    = paramvector
    astroparams = zeus21.Astro_Parameters(CosmoParams,epsstar=10**log10eps, 
        Mc=10**log10Mc,alphastar=alphastar, betastar=betastar, sigmaUV=sigmaUV,
        dlog10epsstardz=dlog10eps,dlog10Mcdz=dlog10Mc,dalphastar=dalphastar, dbetastar=dbetastar, dsigmaUV=dsigmaUV,accretion_model=accretion_model) 
    
    return astroparams


def UVLF_wrapper(zcenters, zwidths, MUVcenters, MUVwidths, paramvector,accretion_model=accretion_model_DEF,CosmoParams = CosmoParams,HMFintclass = HMFintclass,RETURNBIAS=False):
    'Computes and returns the UVLF at z=zcenters, with width zwidths, in bins centered at MUVcenters with width MUVwidths'
    'Output is PhiUV, dependson astro parameter inputs paramvector'
    astroparams = param_wrapper(paramvector,accretion_model=accretion_model)
    UVLFs_std = zeus21.UVLFs.UVLF_binned(astroparams,CosmoParams,HMFintclass,zcenters,zwidths,MUVcenters,MUVwidths,RETURNBIAS=RETURNBIAS)
    return UVLFs_std
```

```python
def log_split_Normal(ydat,theory,pluserr,minuserr):
    'log(Gaussian) for an asymmetric errorbars plus and minus. Unnormalized as usual, see Split_normal_distribution in wiki for normalization'
    _diff = (theory-ydat)
    return -np.sum(_diff**2/(2.0) * (np.heaviside(-_diff,0.5)/minuserr**2 + np.heaviside(_diff,0.5)/pluserr**2) )


fakecenters=np.array([0.0,0.0])
pluserr = np.array([1.0,1.0])
negerr = np.array([0.1,0.01])

fakeres = np.array([1.0,1.0]) #++, etc.
print(log_split_Normal(fakecenters,fakeres,pluserr,negerr))
fakeres = np.array([1.0,-1.0])
print(log_split_Normal(fakecenters,fakeres,pluserr,negerr))
fakeres = np.array([-1.0,1.0])
print(log_split_Normal(fakecenters,fakeres,pluserr,negerr))
fakeres = np.array([-1.0,-1.0])
print(log_split_Normal(fakecenters,fakeres,pluserr,negerr))

```

```python
def log_prior_astro(paramvector):
    'Modify this to whichever params youre varying'

    log10epsstar, log10Mcstar, alphastar, betastar, sigmaUV = paramvector[0:5]
    derivatives = np.array(paramvector[5:])
    
    logprior = 0.0
    if (log10epsstar>1.0 or log10epsstar<-5.0):
        logprior -= np.inf  
    elif (log10Mcstar>15 or log10Mcstar<10):
        logprior -= np.inf  
    elif (alphastar<0.0 or alphastar>3.0):
        logprior -= np.inf
    elif (betastar>0 or betastar<-3.0):
        logprior -= np.inf
    elif (sigmaUV<0.1 or sigmaUV>1.5):
        logprior -= np.inf
    if(np.any(derivatives>0.7) or np.any(derivatives<-0.7)):
        logprior -= np.inf  

    logprior += -0.5*np.sum((derivatives-_priord_centers)**2/(_priord_err**2))
    
    return logprior
  
    
def log_like_astro(paramvector, datavector):
    'Sample flat in eps, alph, beta, and log10Mc. datavector has all the z and UVLF data'
    
    #now bin it appropriately at each z -- data part
    loglike_curr = 0.0

    for dataarrayz in datavector:
        #do each z at a time. Eg datHSTz4=[3.8, mags_z4,phi_z4,err_z4,errx_z4]
        zdat = dataarrayz[0]
        zerr = dataarrayz[1]
        xdat = dataarrayz[2]
        ydat = dataarrayz[3]
        yerrplus = dataarrayz[4] 
        yerrminus = dataarrayz[5] 
        xerr = dataarrayz[6]                
        
        yerrplus = np.fmax(yerrplus, ydat*MINRELERROR)
        yerrminus = np.fmax(yerrminus, ydat*MINRELERROR)
                    
                    
        uvlftheory = UVLF_wrapper(zdat,zerr,xdat,xerr, paramvector)
        
        #asymmetric Gaussian
        loglike_curr += log_split_Normal(ydat,uvlftheory,yerrplus,yerrminus)
        ####OLD:loglike_curr += -np.sum( (ydat - uvlftheory)**2/(2.0 * ((yerrplus+yerrminus)/2.0)**2) ) #assumed Gaussian, to be revisited.
    
    return loglike_curr


def log_prob_astro(paramvector, datavector):
    
    lprior=log_prior_astro(paramvector)
    if (lprior > -np.inf): #only run if in prior range. avoid weird behavior for negative logs etc
        lpost=log_like_astro(paramvector, datavector)
    else:
        lpost = 0.0 #doesn't matter, added to -inf
    return lprior + lpost
```

Now that we know it works let's run the MCMC!

```python
ndim = len(params1)
nwalkers = 2*ndim

p0astro = (1 + np.random.rand(nwalkers, ndim) * 2) * params1
#trim p0 of walkers that are outside of prior range
for ip,pval in enumerate(p0astro):
    priorval = log_prior_astro(p0astro[ip]) 
    while(priorval<-1000.0):
        p0astro[ip] = (1 + np.random.rand(ndim) * 2) * params1
        priorval = log_prior_astro(p0astro[ip])   


#no lensing here
sampler_hst_astro = emcee.EnsembleSampler(nwalkers, ndim, log_prob_astro, args=[datHST])
state_hst_astro = sampler_hst_astro.run_mcmc(p0astro, 100)
sampler_hst_astro.reset()      
```
