Calculate distance to a certain redshit & the limiting absolute magnitude at that redshift (given the limiting apparent magnitude)

```python
from scipy.integrate import quad
import zeus21
from astropy.cosmology import Planck18
```

$d = c*t(a)$ \
$d= \int \frac{c}{aH(a)}da = (1+z)*c*\int_0^z dz'/H(z')$

```python
UserParams = zeus21.User_Parameters()
CosmoParams_input = zeus21.Cosmo_Parameters_Input(zmin_CLASS=0.0)
CosmoParams,ClassyCosmo, CorrFclass, HMFintclass =  zeus21.cosmo_wrapper(UserParams, CosmoParams_input) 
```

```python
zeus21.cosmology.Hub(CosmoParams, 0)
```

```python
def d_L(z, CosmoParams):
    zs = np.linspace(0, z, 100)
    Hs = zeus21.cosmology.Hub(CosmoParams, zs)
    c = 3e5 # speed of light in km/s
    integral = (1+z) * c * np.trapz(1/Hs, zs)
    return integral
```

$M-m = -5(\log_{10}(D/10pc))$

```python
def absolute_magnitude(apparent_mag, z, CosmoParams):
    D = d_L(z, CosmoParams)*1e6 # Multiply by 1e6 to convert to pc

    return (-5*np.log10(D/10)) + apparent_mag
```

```python
absolute_magnitude(26.5, 4, CosmoParams)
```

```python
def V(zcenter, z_err, survey_area):
    sky_area = 4*np.pi * (180/np.pi)**2 # 4pi steradians to degrees
    fraction = survey_area / sky_area
    z1, z2 = zcenter - z_err, zcenter + z_err
    return (Planck18.comoving_volume(z2) - Planck18.comoving_volume(z1))*fraction
```

```python
# Roman survey volume
V(4, 0.5, 2000)
```
