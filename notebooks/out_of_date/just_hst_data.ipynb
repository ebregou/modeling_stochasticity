Run chain with just HST data to see what is preferred

```python
from escripts import eMCMC
from escripts import edata
from escripts import eplots
```

```python
file_names = ['/Users/eb35267/Desktop/code/home/data/Bouwens2021_fixed.txt']
data_labels = ['Bouwens+21']
sorted_data = edata.get_sorted(file_names, data_labels)
```

```python
params = eMCMC.build_param_data({})
```

```python
my_UVLF = eMCMC.UVLF(sorted_data, params, backend_filename = '/Users/eb35267/Desktop/code/home/data/just_HST.h5')
```

```python
my_UVLF.run_MCMC(Nsteps = 25000)
```

```python
corner = eplots.make_corner(my_UVLF)
```

```python
evolv = eplots.evolving_UVLF_fit(my_UVLF)
```
