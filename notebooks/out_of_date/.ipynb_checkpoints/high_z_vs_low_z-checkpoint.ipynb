```python
def plot_evolving_sigma(bf_low_z, bf_high_z, z_pivot = 8, Mh_pivot = 12):

    fig, ax = plt.subplots()

    Mhs = np.linspace(9, 13, 5)
    cmap = LinearSegmentedColormap.from_list("custom", [eplots.red, eplots.navy])
    norm = BoundaryNorm(np.arange(min(Mhs), max(Mhs)+1, 1), cmap.N)
    
    low_z = np.linspace(0, 10, 100)
    high_z = np.linspace(10, 14, 100)
    sigma0_lowz, dsigdz_lowz, dsigdM_lowz = bf_low_z[-3], bf_low_z[-2], bf_low_z[-1]
    sigma0_highz, dsigdz_highz, dsigdM_highz = bf_high_z[-3], bf_high_z[-2], bf_high_z[-1]

    for Mh in Mhs:
        low_z_sigs = sigma0_lowz + (dsigdz_lowz*(low_z-z_pivot)) + (dsigdM_lowz*(Mh-Mh_pivot))
        high_z_sigs = sigma0_highz+ (dsigdz_highz*(high_z-z_pivot)) + (dsigdM_highz*(Mh-Mh_pivot))
        low_z_sigs = low_z_sigs.clip(min = 0.3)
        high_z_sigs = high_z_sigs.clip(min = 0.3)

        frac = (Mh - Mhs.min()) / (Mhs.max() - Mhs.min())
        color = cmap(frac)
        ax.plot(low_z, low_z_sigs, c = color, ls = '-')
        ax.plot(high_z, high_z_sigs, c = color, ls = '-')

    ax.axhline(0.3, color = eplots.turquoise, ls = '--', label = r'min($\sigma_{\rm{UV}} = 0.3$)', zorder = 0)


    # Color bar things:
    r'$\sigma_{\rm{UV}}$'
    

    ax.set_xlabel(r'$z$')
    ax.set_ylabel(r'$\sigma_{\rm{UV}}$')
    ax.legend()

    return fig
```

```python
evolv_sig = plot_evolving_sigma(lzbf, hzbf)
```
