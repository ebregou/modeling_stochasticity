```python
from escripts import edata
from escripts import eMCMC
from escripts import eplots
```

```python
high_z_files = ['/Users/eb35267/Desktop/code/home/data/Donnan24_GAL.txt',
             '/Users/eb35267/Desktop/code/home/data/Mcleod_24.txt']
data_labels = ['Donnan+24', 'McLeod+24']
high_z_data = edata.get_sorted(high_z_files, data_labels)
```

```python
backend_file = '/Users/eb35267/Desktop/code/home/data/high_z.h5'
```

```python
params = eMCMC.build_param_data({})
```

```python
high_z_UVLF = eMCMC.UVLF(high_z_data, params)
```

```python
high_z_UVLF.run_MCMC(Nsteps = 40000)
```

```python
samples, best_fit, labels = high_z_UVLF.get_fit()
```

```python
lowers, uppers = np.zeros(len(samples[0])), np.zeros(len(samples[0]))
for i in range(len(samples[0])):
    lowers[i], uppers[i] = np.percentile(samples[:, i], [16, 84])
```

```python
def make_table(best_fits, param_labels, fit_labels, bounds = None):
    """
    Make a table to compare best fit values of parameters
    Inputs:
        best_fits [list of lists]: list of best fit parameters
        param_labels [list of strs]: names of parameters (rows of the table)
        fit_labels [list]: names of each type of fit (columns of the table)
    Outputs:
        dataframe table with labeled parameters for comparison
    """
    if bounds is not None: # Format the parameters with +/- 
        df_fill = [f'${bf}^{+(bound[1]-bf)}_{(bound[0]-bf)}$' for bf, bound in zip(best_fits, bounds)]
    else:
        df_fill = best_fits
    df = pd.DataFrame(df_fill, columns = param_labels)
    df.index = fit_labels
```
