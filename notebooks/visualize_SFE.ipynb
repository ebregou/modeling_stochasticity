Visualize what parameters mean for SFE shape

```python
alpha = 0.71
beta = 0.12
logeps = -1.42
logMc = 11.75
```

```python
Mh = np.logspace(8, 17, 100)
```

```python
sfe = 10**logeps/((Mh/10**logMc)**-alpha + (Mh/10**logMc)**-beta)
```

```python
plt.plot(Mh, sfe)
plt.xscale('log')
plt.yscale('log')
plt.ylabel('SFE')
plt.xlabel(r'$M_h$')
```
