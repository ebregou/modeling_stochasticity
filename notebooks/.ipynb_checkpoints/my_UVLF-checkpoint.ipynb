```python
def calc_M_dot(M_h, params, z = 0):
    return 25.3*(M_h/1e12)**(1.1)*(1+(1.65*z))*np.sqrt((params['Omega_m']*(1+z)**3)+params['Omega_L'])
```

```python
def calc_SFE(M_h, eps_0 = 0.1, M_0 = 1e12, alpha = 0.6, beta = 0.5):
    return 2*eps_0/((M_h/M_0)**-alpha+(M_h/M_0)**beta)
```

```python
# SFE = calc_SFE(M_halo)
# fig, ax = plt.subplots()
# ax.loglog(M_halo, SFE)
# ax.set_xlabel(r'Mass [$M_{\odot}$]')
# ax.set_ylabel(r'SFE')
# ax.set_title('SFE')
```

```python
def calc_SFR(M_h, params, z=0):
    fb = params['Omega_b']/params['Omega_m']
    M_dot = calc_M_dot(M_h, params, z)
    eps_star = calc_SFE(M_h)
    return fb*M_dot*eps_star
```

```python
# sfrs = calc_SFR(M_halo, params, z)
# fig, ax = plt.subplots()
# ax.loglog(M_halo, sfrs)
# ax.set_xlabel(r'Mass [$M_{\odot}$]')
# ax.set_ylabel(r'SFR[$M_{\odot}/\rm{yr}^{-1}$]')
# ax.set_title('SFR')
```

```python
def calc_L_UV(M_h, params, z=0, Kappa_UV = 1.15e-28):
    SFR = calc_SFR(M_h, params, z)
    return SFR/Kappa_UV

def calc_M_UV(M_h, params, z=0):
    L_UV = calc_L_UV(M_h, params, z)
    M_UV = 51.63-(2.5*np.log10(L_UV)) # Same as Zeus
    return M_UV

def M_to_L(M_UV):
    return 10**(0.4*(51.63-M_UV))
```

```python
L_UV = calc_L_UV(M_halo, params, z)
M_UV = calc_M_UV(M_halo, params, z)
```

```python
def calc_M_UV_dust(M_halo, params, z=0):
    M_UV = calc_M_UV(M_halo, params, z)
    M_UV_obs = (M_UV - 6.35)/1.34
    M_UV_obs = np.where(M_UV_obs > M_UV, M_UV_obs, M_UV) # Make sure you're not making faint galaxies brighter

    return M_UV_obs
```

```python
M_UV_dust = calc_M_UV_dust(M_halo, params, z)
fiducial_uvlf_dust = np.log10(hmf*np.abs(np.gradient(M_halo, M_UV_dust))) # Must take the absolute value of dM/dM_UV
```

```python
def interpolate(x0, y0, x1):
    # Make sure the range you're interpolating is defined for both
    defined_range = np.where((x1 >= np.min(x0)) & (x1 <= np.max(x0)))[0] # This is the indices in x1 where x0 is defined. Use it to cut
    # x1 and y1 when you're plotting to make sure everything is consistent
    try: 
        interpolater = interp1d(x0, y0)
        return interpolater(x1[defined_range]), defined_range
    except: # If your x0 is strictly DECREASING, this will work by switching the order of everything
        interpolater  = interp1d(x0[::-1], y0[::-1])
        return interpolator(x1[defined_range])[::-1], defined_range
```

# Set up MCMC

```python
class UVLF_MCMC:
    def __init__(self, lowers, uppers, z = 0, Mmin = 9, Mmax = 17, dlog10m = .05, hmf_model = 'Tinker10'):
        self.params = cosmo_calc.get_params()
        self.z = z
        my_hmf = halomassfunction.hmf.MassFunction(Mmin = Mmin, Mmax = Mmax, dlog10m = dlog10m, hmf_model = hmf_model, z = self.z)
        self.M_h = my_hmf.m/params['h']
        self.hmf = my_hmf.dndm*params['h']**4
        self.lowers = lowers
        self.uppers = uppers
        self.labels = labels
        self.start_vals = start_vals
        self.nwalkers = nwalkers
        self.N = N # Length of chain
        
    # Gaussian log-likelihood
    def log_likelihood(self, theta, x, y, yerr, bw):
        """
        theta [1darr]: list of the current values of the parameters that need to be estimated
        x, y [1darr, 1darr]: the datapoints to fitt
        yerr [1darr]: error on the datapoints
        """
        model = self.UVLF_model(x, bw, *theta)
        if np.isnan(model).any():
            print(model)
            print(theta)
        return -0.5 * np.sum(((y - model) / yerr)**2)
    
    # Log uniform prior
    def log_prior(self, theta):
        if all(lower < t < upper for t, lower, upper in zip(theta, self.lowers, self.uppers)):
            return 0.0  # log(1)
        return -np.inf  # log(0)
    
    # Combined log-posterior
    def log_probability(self, theta, x, y, yerr, bw):
        lp = self.log_prior(theta)
        if not np.isfinite(lp):
            return -np.inf
        return lp + self.log_likelihood(theta, x, y, yerr, bw)

    def interpolate(self, x0, y0, x1):
        try: 
            interpolater = interp1d(x0, y0, fill_value = 'extrapolate')
            return interpolater(x1)
        except: # If your x0 is strictly DECREASING, this will work by switching the order of everything
            interpolater  = interp1d(x0[::-1], y0[::-1], fill_value = 'extrapolate')
            return interpolator(x1)[::-1]

    def UVLF_model(self, x, bw, log_eps_0, alpha, beta, log_M_0):
        M_dot = self.calc_M_dot()
        SFE = self.calc_SFE(10**log_eps_0, alpha, beta, 10**log_M_0) # We're sampling eps_0 and M_0 in logspace, so we need to unlog now
        SFR = self.calc_SFR(M_dot, SFE)
        M_UV = self.calc_M_UV(SFR)
        M_UV_dust = self.calc_M_UV_dust(M_UV)
        uvlf = np.log10(self.hmf*np.abs(np.gradient(self.M_h, M_UV_dust)))
        
        if bw is None:
            model = self.interpolate(M_UV_dust, uvlf, x)
            model[np.isnan(model)] = -500 # Put in a value where otherwise a nan was returned
            return model
        else:
            binned_uvlf = np.log10(self.binned_UVLF(M_UV_dust, 10**uvlf, x, bw)) # Need the not log version to get the binned UVLF
            binned_uvlf[np.isnan(binned_uvlf)] = -500 # Put in a value where otherwise a nan was returned
            return binned_uvlf
        
    def calc_M_dot(self):
        return 25.3*(self.M_h/1e12)**(1.1)*(1+(1.65*self.z))*np.sqrt((self.params['Omega_m']*(1+z)**3)+self.params['Omega_L'])

    def calc_SFE(self, eps_0, alpha, beta, M_0):
        return 2*eps_0/(((self.M_h/M_0)**(-alpha))+(self.M_h/M_0)**beta)

    def calc_SFR(self, M_dot, SFE):
        fb = self.params['Omega_b']/self.params['Omega_m']
        return fb*M_dot*SFE

    def calc_M_UV(self, SFR, Kappa_UV = 1.15e-28):
        L_UV = SFR/Kappa_UV
        M_UV = 51.63-(2.5*np.log10(L_UV))
        return M_UV

    def calc_M_UV_dust(self, M_UV):
        M_UV_obs = (M_UV - 6.35)/1.34
        M_UV_obs = np.where(M_UV_obs > M_UV, M_UV_obs, M_UV) # Make sure it doesn't boost the brightness of faint galaxies
    
        return M_UV_obs 

    def binned_UVLF(self, M_UV, phi_UV, M0s, delta_Ms):
        """
        phi_UV has to be phi_UV not log(phi_UV)
        """
        binned_phi = []
        for M0, delta_M in zip(M0s, delta_Ms): # Iterate through each bin
            W_UV = ((M_UV > M0 - delta_M/2) & (M_UV < M0 + delta_M/2)).astype(float) # Array of 1s and 0s corresponding to whether that M_UV
            # falls within the given bin
            phi_bin = np.trapz(W_UV * phi_UV, np.abs(M_UV)) / delta_M  # Integrate the UVLF over this bin. Dividing by delta_M normalizes 
            # Take the absolute value of M_UV so that everything is positive
            binned_phi.append(phi_bin)
        
        return np.array(binned_phi)
```

```python
def run_MCMC(data_x, data_y, yerr, bw, lowers, uppers, start_vals, z = 0, nwalkers = 30, N = 5000):
    """
    If you don't want a binned one, just set bw to None.
    """
    ndim = len(start_vals)
    start = start_vals + 1e-4*np.random.randn(nwalkers, ndim) # Start each of the walkers in a slightly different place
    my_MCMC = UVLF_MCMC(lowers, uppers, z)
    
    sampler = emcee.EnsembleSampler(nwalkers, ndim, my_MCMC.log_probability, args=(data_x, data_y, yerr, bw))
    sampler.run_mcmc(start, N)
    samples = sampler.get_chain(discard=100, flat = True)
    
    return my_MCMC, samples
```

```python
def plot_corner(samples, labels, Plotting, true_vals = None):
    if true_vals is None:
        corner_plot = corner.corner(samples, labels=labels, color = Plotting.red) 
    else:
        corner_plot = corner.corner(samples, labels=labels, color = Plotting.red, truths = true_vals, truth_color=Plotting.turquoise)
        
    return corner_plot
```

```python
def MCMC_vs_data(data_x, data_y, yerr, bw, all_samples, my_MCMC, data_label, Plotting, nsamples = 100, truths = None):
    """
    my_MCMC [UVLF_MCMC object]
    """
    fig, ax = plt.subplots()
    inds = np.random.randint(len(all_samples), size=nsamples) # Choose nsamples from the chain
        
    for ind in inds:
        sample = all_samples[ind]
        ax.plot(data_x, my_MCMC.UVLF_model(data_x, bw, *sample), alpha=0.03, color = Plotting.red, zorder = 0)

    if truths is not None: # Plot the true curve if one is known. Otherwise, plot the best fit
        grid = np.linspace(np.min(data_x), np.max(data_x), 100) # Make a continuous grid if there is a true model available
        ax.plot(grid, my_MCMC.UVLF_model(grid, bw, *truths), color = Plotting.red, label = 'true UVLF', zorder = 1)
    else: 
        best_fit = mode(all_samples, axis = 0).mode # Mode seems to be a better statistics than average
        #best_fit = np.average(all_samples, axis = 0)
        ax.plot(data_x, my_MCMC.UVLF_model(data_x, bw, *best_fit), color = Plotting.red, label = 'best fit (mode)', zorder = 1)
    
    ax.plot([0], [0], color = Plotting.red, alpha = 0.5, label = 'sampled fits') # Create the label for the samples
    ax.errorbar(data_x, data_y, yerr, fmt = ".", capsize=0, markersize = 10, color = Plotting.navy, label = data_label)
    
    ax.invert_xaxis()
    ax.legend()
    ax.set_ylim(np.min(data_y)-1, np.max(data_y)+1)
    ax.set_xlim(np.max(data_x)+0.25, np.min(data_x)-0.25)
    
    ax.set_xlabel(r'$M_{\rm{UV}}$ [mag]')
    ax.set_ylabel(r'$\log_{10}(\phi_{\rm{UV}}$ [$\rm{mag}^{-1} \rm{Mpc}^{-3}$])')
    ax.set_title(f'UVLF data vs. MCMC model, $z = {my_MCMC.z}$')

    return fig, best_fit
```
